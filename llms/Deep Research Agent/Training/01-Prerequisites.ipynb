## NVIDIA API Key

The NVIDIA API key authenticates your access to NVIDIA's NGC (NVIDIA GPU Cloud) services, letting you pull GPU-optimized containers, access pre-trained AI models, and interact with NVIDIA's cloud infrastructure programmatically. The key is unique to your account and can be configured with specific permissions—you control which NGC services it can access and set expiration times ranging from one hour to no expiration.

In this course, we will use the key to authenticate all model endpoint API calls to the NVIDIA-hosted endpoints. These calls are free with no cap on usage; however note that there are developer-friendly rate limits in place when using any model in the API catalog. 

### Generate your API Key - NVIDIA NGC

The recommended way to generate this key is to do so through NVIDIA's NGC platform. This enables more fine-grained control over permissions, access, and expiration. 

1. Go to [NVIDIA NGC](https://catalog.ngc.nvidia.com/) and log in / sign up for your NGC account
2. Choose an NGC org from the profile dropdown menu. Ensure the org has API access, eg. [NVIDIA Developer Program](https://developer.nvidia.com/developer-program), etc.
3. Select **Setup** from the profile dropdown menu
4. Select **Generate API Key**
5. Select **Generate Personal Key**
6. Name the key and set the expiration if desired
7. For **Services Included**, select:
   - ✅ NGC Catalog
   - ✅ Public API Endpoints
8. Select **Generate Personal Key** when finished and save / store the key in a safe place for later use.



Your key should start with ``nvapi-...`` and have access to the same model endpoints your NGC org has access to. 

## Tavily API Key

The Tavily API key is an authentication token that provides access to [Tavily's](https://www.tavily.com/) specialized search engine API designed specifically for AI developers. It allows developers to perform advanced web searches that return structured results including titles, content snippets, URLs, and even related images when requested. Unlike traditional search engines, Tavily is optimized for AI applications, delivering factual, real-time results specifically formatted for consumption by language models.

### Generate your API Key

1. Go to [Tavily's webpage](https://www.tavily.com/) and create a free account.
2. On the homepage, click the plus sign next to **API Keys** and generate a key.
3. Save and store this key for future use.



The Tavily API key should begin with ``tvly-...`` and comes with 1000 free credits to use per month.
